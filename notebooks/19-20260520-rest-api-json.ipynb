```python
import os, requests
from dotenv import load_dotenv

load_dotenv()
KEY= os.environ["KOREAN_DICT_KEY"]
KEY

params = {
    "key": KEY, "q": "언어", "req_type": "json",
    "num": 10, "start": 1, "type1": "word",
}

r = requests.get("https://opendict.korean.go.kr/api/search",params=params,timeout=10)
r.raise_for_status()
```
