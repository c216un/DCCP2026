```python
df = pd.DataFrame(
    {
        "이름": ['강은수', '조재영', '이한솔', '뽀로로'],
        "학과": ['언어학', '국문학', '언어학', '역사학'],
        "학년": [2, 3, 1, 4],
        "국어": [88, 72, 95, 60],
        "수학": [75, 88, 82, 50],
        "영어": [90, 79, 85, 70]
    }
)

print(df)
```

```python
df['평균'] = (df['국어'] + df['수학'] + df['영어']) / 3
df
```

```python
df['합격'] = df['평균'] >= 70
df
```

```python
def grade(score:int) -> str:
    if score >= 90:
        return 'A'
    if score >= 80:
        return 'B'
    if score >= 70:
        return 'C'
    if score >= 60:
        return 'D'
    return 'F'

df['등급'] = df['국어'].apply(grade)
df
```

```python
df.sort_values('평균', ascending=False)
```
