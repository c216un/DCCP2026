pd.merge는 두 DataFrame을 공통 키(on='작가')를 기준으로 합친다. how='left'이므로 왼쪽 표인 works2의 행 수(11행)가 그대로 유지되고, authors에서 해당 작가의 생년·몰년 열이 붙는다. shape를 출력해 merge 후에도 행 수가 변하지 않았음을 확인한다. 

확인 질문: authors에만 있는 '염상섭'은 full에 포함되지 않는다. how='left'는 왼쪽 표(works2)의 행만 기준으로 삼고 오른쪽에서 매칭되는 키가 있을 때만 열을 결합하므로, 오른쪽에만 존재하는 행은 결과에 나타나지 않기 때문이다.
